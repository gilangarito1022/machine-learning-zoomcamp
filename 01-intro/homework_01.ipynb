```python
# Median value of horsepower before filling missing values
median_hp_carfuel_before = data_carfuel['horsepower'].median()
print(median_hp_carfuel_before)
```

```python
# the most frequent value of the same horsepower column.
mode_hp_carfuel = data_carfuel['horsepower'].mode()[0]
print(mode_hp_carfuel)
```

```python
# Fill missing value in the horsepower column with the most frequent value
data_carfuel['horsepower'] = data_carfuel['horsepower'].fillna(mode_hp_carfuel)

# Median value of horsepower after filling missing values
median_hp_carfuel_after = data_carfuel['horsepower'].median()
print(median_hp_carfuel_after)
```

```python
# Sum of weights
# 1. Select all the cars from Asia
asia_cars = data_carfuel[data_carfuel['origin'] == 'Asia'].reset_index(drop = True)

# 2. Select only columns vehicle_weight and model_year
asia_cars_subset = asia_cars[['vehicle_weight', 'model_year']].head(7)

# 3. Select the first 7 values
asia_cars_subset
```

```python
# 4. Get the underlying NumPy array. Let's call it X
X = asia_cars_subset.values
X
```

```python
# 5. Compute matrix-matrix multiplication between the transpose of X and X.
# To get the transpose, use X.T. Let's call the result XTX
X_transpose = X.T
X_transpose
```

```python
# Compute matrix-matrix multiplication between the transpose of X and X.
XTX = X_transpose.dot(X)
XTX
```

```python
import numpy as np

# 6. Invert XTX
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
# 7. Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200]
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
# 8. Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w
w_half = XTX_inv.dot(X_transpose)
w = w_half.dot(y)
print("Sum of weights:", w.sum().round(3))
```
